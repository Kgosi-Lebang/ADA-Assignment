# Data Dictionary

| Variable Name | Meaning | Data Type | Notes |
|---|---|---|---|
| order_id | Unique order identifier | Integer | Primary key |
| user_id | Customer identifier | Integer | Links customers to orders |
| order_number | Sequence number of customer orders | Integer | Shows purchase order |
| order_dow | Day of the week order was placed | Integer | 0–6 |
| order_hour_of_day | Hour order was placed | Integer | 0–23 |
| days_since_prior_order | Days since previous order | Float | Contains missing values |
| product_id | Unique product identifier | Integer | Primary key |
| product_name | Name of product | String | Product description |
| aisle_id | Aisle identifier | Integer | Links to aisles |
| department_id | Department identifier | Integer | Links to departments |

# Table Relationships

The Instacart dataset is relational and consists of multiple linked tables.

## Orders Table
- order_id is the primary key.
- user_id identifies customers.

## Products Table
- product_id is the primary key.
- aisle_id and department_id classify products.

## Order Products Table
- order_id links to the orders table.
- product_id links to the products table.

The order_products table acts as a bridge table connecting customer orders with purchased products.

# Business Understanding

The Instacart Market Basket Analysis dataset represents an online grocery retail business where customers place orders for various products. The dataset captures customer purchasing behaviour, product information, and relationships between orders and items purchased.

This data can help businesses understand customer buying patterns and improve operational decision-making. For example, analysts can identify which products are frequently purchased together, which products are reordered most often, and which times of day experience the highest order activity.

The dataset can support several important business decisions. Marketing teams can use the data to create personalised product recommendations and targeted promotions. Inventory managers can use purchasing trends to improve stock planning and reduce shortages. Management can also use customer behaviour analysis to improve customer retention and increase sales performance.

From an analytics perspective, the dataset supports descriptive, predictive, and prescriptive analysis. Descriptive analytics can summarise customer ordering behaviour and popular products. Predictive analytics can estimate future purchasing behaviour or predict product reordering patterns. Prescriptive analytics can recommend business actions such as product placement strategies, recommendation systems, or promotional campaigns.

Overall, the dataset provides a realistic example of how large-scale business data can be analysed to support data-driven decision-making in retail organisations.
